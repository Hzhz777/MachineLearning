## 1. 读取数据

运行下面的单元格，然后开始独立练习。

```python
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import SGDRegressor, LinearRegression
from sklearn.metrics import mean_squared_error

train_data = np.loadtxt("train.txt")
test_data = np.loadtxt("test.txt")

X_train = train_data[:, :4]
y_train = train_data[:, 4]
X_test = test_data[:, :4]
y_test = test_data[:, 4]

print("训练集：", X_train.shape, y_train.shape)
print("测试集：", X_test.shape, y_test.shape)
```

## 2. 用 StandardScaler 标准化

完成训练集和测试集的特征标准化；预处理所需的信息只能从训练集获得。

输出标准化后训练集每列的均值和标准差，检查结果。


```python
scaler = StandardScaler()
scaler.fit(X_train)
X_train_scaler  = scaler.transform(X_train)
```

## 3. 用 SGDRegressor 训练

创建并训练模型，设置随机种子为 42，保证重复运行时便于比较。其他配置由你决定。

输出模型学到的权重、截距以及实际训练轮数。若出现未收敛提示，先阅读提示，再检查训练设置。


```python
np.random.seed(42)
```

```python
model = SGDRegressor(max_iter=1000,random_state=42)
```

```python
model.fit(X_train_scaler,y_train)
```

```python
print(f'权重w:{model.coef_}')
print(f'截距b:{model.intercept_}')
print(f'实际训练轮数：{model.n_iter_}')
```

## 4. 使用模型预测并评估

用模型的预测接口得到训练集与测试集的预测结果。

计算并输出训练成本、测试成本，以及前 5 个测试样本的真实值和预测值。成本定义沿用练习 1 的 $J=\frac{1}{2m}\sum_i(\hat y_i-y_i)^2$，便于比较。

不要根据测试集成绩反复选择参数；它用于最终评估。


```python
X_test_scaled = scaler.transform(X_test)
```

```python
print('real:',y_test[:5])

print('predict:',model.predict(X_test_scaled[:5]))
```

```python
print(f'J_cost is:{mean_squared_error(y_test,model.predict(X_test_scaled))/2}')
```

## 5. 用 LinearRegression 再完成一次

使用相同的训练数据和标准化结果，换成 LinearRegression 训练、预测。

输出权重、截距、训练成本、测试成本和前 5 个测试样本的预测值。请另起模型变量名，方便和前面的模型比较。
