```python
import numpy as np
x_train = np.array([1.0, 1.5, 2.0, 2.5, 3.0])
y_train = np.array([110., 155., 210., 255., 310.])
```

```python
import random
```

```python
import copy
```

```python
# y = wx + b
random.seed(42)
w = random.randint(1,10)
b = random.randint(1,10)
```

goal:最小化$J = \sum_{i=1}^{m} (f_{w,b}(x_i)-y_i)^2$

方法：梯度下降

$w = w - \alpha \times \frac{\partial J}{\partial w}$

$b = b - \alpha \times \frac{\partial J}{\partial b}$

```python
alpha = 0.1
```

```python
def compute_loss(x,y,w,b):
    m = x.shape[0]
    total_loss = 0
    for i in range(m):
        total_loss += (w*x[i]+b-y[i])*(w*x[i]+b-y[i])
    return total_loss/(2*m)
```

```python
def compute_gradient(x,y,w,b):
    m = x.shape[0]
    dj_dw = 0
    dj_db = 0
    for i in range(m):
        dj_dw += (w*x[i] + b - y[i]) * x[i]
        dj_db += (w*x[i] + b - y[i])
    dj_dw = dj_dw / m
    dj_db = dj_db / m
    return dj_dw,dj_db
```

```python
def gradient_descend(x,y,w_in,b_in,iteration,loss_function,gradient_function):
    m = x.shape[0]
    w = copy.deepcopy(w_in)
    b = copy.deepcopy(b_in)
    J_history = [loss_function(x,y,w,b)]
    p_history = [(w,b)]
    for i in range(iteration):
        dj_dw,dj_db = gradient_function(x,y,w,b)
        w = w - alpha * dj_dw
        b = b - alpha * dj_db
        j = loss_function(x,y,w,b)
        J_history.append(j)
        p_history.append((w,b))
        if i % np.ceil(iteration/10) == 0:
            print('='*50)
            print(f'iteration {i}:now the loss is equal to {j}')
            print(f'iteration {i}:the parameters is ({w},{b})')
            print('='*50)
    return w,b
        
        
        
        
    
```

```python
w,b = gradient_descend(x_train,y_train,w,b,10000,compute_loss,compute_gradient)
```

```python
print(f'the result of forecast is {w*1.8 + b :.2f}')
```
